```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_dirty_sample.csv")
```

```python
df.columns
df[["product_name", "category", "unit_price", "total_amount"]].head()
df["unit_price"].info()
```

```python
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
```

```python
df["unit_price"].isnull().sum()
df[df["unit_price"].isnull()]
df[df["unit_price"] < 0]
```

```python
max_price = df["unit_price"].max()
print("Highest product price:", max_price)
```

```python
highest_price_record = df[df["unit_price"] == max_price]
highest_price_record
```

```python
highest_price_record[
    ["order_id", "order_date", "customer_id", "product_name",
     "category", "quantity", "unit_price", "total_amount",
     "payment_method", "order_status", "city"]
]
```

```python
print("Maximum price:", df["unit_price"].max())
print("Price in selected record:", highest_price_record["unit_price"].iloc[0])
print("Product name:", highest_price_record["product_name"].iloc[0])
print("Category:", highest_price_record["category"].iloc[0])
print("Order ID:", highest_price_record["order_id"].iloc[0])

assert highest_price_record["unit_price"].max() == max_price
```

```python
print("Total records:", len(df))
print("Missing prices:", df["unit_price"].isnull().sum())
print("Negative prices:", (df["unit_price"] < 0).sum())
print("Highest valid numeric price:", df["unit_price"].max())
```
