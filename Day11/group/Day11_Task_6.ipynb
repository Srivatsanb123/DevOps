```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_dirty_sample.csv")
```

```python
df[["city", "quantity", "unit_price"]].head()
df[["city", "quantity", "unit_price"]].info()
df[["city", "quantity", "unit_price"]].isnull().sum()
```

```python
df["city"] = df["city"].astype("string").str.strip().str.title()
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce")
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
```

```python
df["city"] = df["city"].replace("", pd.NA)
df.loc[df["quantity"] <= 0, "quantity"] = np.nan
df.loc[df["unit_price"] < 0, "unit_price"] = np.nan

df[["city", "quantity", "unit_price"]].isnull().sum()
```

```python
df["revenue"] = df["quantity"] * df["unit_price"]
```

```python
df[["order_id", "city", "quantity", "unit_price", "revenue"]].head(10)
df["revenue"].describe()
```

```python
city_revenue = df.groupby("city")["revenue"].sum()
city_revenue
```

```python
city_revenue = city_revenue.sort_values(ascending=False)
city_revenue
```

```python
print("Highest revenue city:", city_revenue.idxmax())
print("Total revenue:", city_revenue.max())
```
